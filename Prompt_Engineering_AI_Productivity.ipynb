```python

import json
import textwrap
from datetime import datetime

print("AI Productivity Toolkit")
print("Project started successfully!")
print("Date:", datetime.now().strftime("%Y-%m-%d"))
```

```python

text = """
Artificial Intelligence is changing many areas such as education,
healthcare, agriculture and business. It helps people automate tasks,
analyze large amounts of data and make better decisions.
"""

prompt = f"""
Summarize the following text in 3 simple bullet points.

TEXT:
{text}
"""

print("ZERO-SHOT PROMPT")
print(prompt)
```

```python

examples = """
Example 1:
Input: "The product is excellent and very useful."
Output: Positive

Example 2:
Input: "The application is slow and difficult to use."
Output: Negative
"""

new_text = "The AI tool is very helpful and easy to use."

prompt = f"""
Classify the sentiment as Positive, Negative, or Neutral.

{examples}

Input: "{new_text}"
Output:
"""

print("FEW-SHOT PROMPT")
print(prompt)
```

```python

prompt = """
Role:
You are an AI career assistant.

Context:
The user is an AIML undergraduate student.

Task:
Create a short professional career summary.

Constraints:
- Use simple English.
- Keep it below 80 words.
- Mention AI and Machine Learning.
- Make it suitable for a resume.
"""

print(prompt)
```

```python

def summarization_prompt(text):
    return f"""
You are a professional summarization assistant.

Task:
Summarize the following text in 5 short bullet points.

Rules:
- Keep the important information.
- Use simple English.
- Do not add information that is not present.

Text:
{text}
"""

sample_text = """
Artificial Intelligence is a branch of computer science that enables
machines to perform tasks that normally require human intelligence.
Machine learning allows computers to learn patterns from data.
AI is used in healthcare, education, agriculture, finance and robotics.
"""

print(summarization_prompt(sample_text))
```

```python

def classification_prompt(text):
    return f"""
Classify the following text into one of these categories:

1. AI
2. Education
3. Business
4. Technology
5. Other

Return only the category name.

Text:
{text}
"""

text = "Machine learning algorithms can identify patterns in large datasets."

print(classification_prompt(text))
```

```python

def create_json_profile(name, skills, education, project):
    data = {
        "name": name,
        "skills": skills,
        "education": education,
        "project": project
    }

    return json.dumps(data, indent=4)

profile = create_json_profile(
    "AIML Student",
    ["Python", "C", "SQL"],
    "B.Tech Artificial Intelligence and Machine Learning",
    "Bluetooth Controlled Robotic Arm"
)

print(profile)
```

```python

def email_prompt(topic, receiver):
    return f"""
You are a professional email writing assistant.

Write a polite and professional email.

Receiver: {receiver}
Topic: {topic}

Requirements:
- Add a suitable subject.
- Keep the email short.
- Use professional language.
- End politely.
"""

print(email_prompt(
    "Request for project discussion",
    "Project Guide"
))
```

```python

def resume_prompt(skills, project):
    return f"""
You are a professional resume assistant.

Create a short resume summary for an AIML student.

Skills:
{skills}

Project:
{project}

Requirements:
- Professional tone
- Simple English
- Maximum 80 words
- Highlight AI/ML interest
"""

print(resume_prompt(
    "Python, C, SQL",
    "Bluetooth Controlled Robotic Arm using Arduino UNO"
))
```

```python

def coding_prompt(problem):
    return f"""
You are an AI coding assistant.

Task:
{problem}

Provide:
1. Explanation
2. Python code
3. Expected output
4. Possible errors
5. Suggestions for improvement

Use beginner-friendly Python.
"""

problem = "Create a Python program to calculate the average of three numbers."

print(coding_prompt(problem))
```

```python

def research_prompt(topic):
    return f"""
You are an AI research assistant.

Research Topic:
{topic}

Organize the information under these headings:

1. Introduction
2. Definition
3. Main Features
4. Applications
5. Advantages
6. Limitations
7. Conclusion

Use simple and clear language.
"""

print(research_prompt("Artificial Intelligence in Education"))
```

```python

test_cases = [
    "Artificial Intelligence is used in healthcare.",
    "Python is a popular programming language.",
    "Students use online platforms for learning."
]

for i, text in enumerate(test_cases, start=1):
    print(f"\nTEST CASE {i}")
    print("-" * 40)
    print(classification_prompt(text))
```

```python

prompt_library = {
    "summarization": "Summarize the given text in 5 simple bullet points.",

    "classification": "Classify the given text into the appropriate category.",

    "email": "Write a professional email based on the given topic.",

    "resume": "Improve the given resume content using professional language.",

    "coding": "Explain, generate and debug the requested Python code.",

    "research": "Organize research information using clear headings."
}

for name, prompt in prompt_library.items():
    print(f"\n{name.upper()}")
    print(prompt)
```
